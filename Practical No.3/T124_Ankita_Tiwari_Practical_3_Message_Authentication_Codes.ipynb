```python
print("ANKITA TIWARI")
print("TYCS")
print("T124")
print("Cyber Security Practical No.3")


import hmac
import hashlib

key = b"secretkey"

message = input("Enter message: ")

mac = hmac.new(key, message.encode(), hashlib.sha256).hexdigest()
print("Generated MAC:", mac)

check = input("Enter message to verify: ")

new_mac = hmac.new(key, check.encode(), hashlib.sha256).hexdigest()

if mac == new_mac:
    print("MAC Verified! Message is Authentic.")
else:
    print("Verification Failed! Message has been Changed.")
```
